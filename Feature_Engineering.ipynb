### `Feature Engineering` ---> It is the Process of Creating New and Better Fature for our Model from which we can have better insights or we can make better prediction model.

```python
import pandas as pd
import seaborn as sns
```

```python
train=pd.read_csv("train.csv")
```

```python
test=pd.read_csv("tested.csv")
```

```python
df=pd.concat([train,test])
```

```python
df['Family Size']=df['SibSp'] + df['Parch'] + 1
```

```python
df.head()
```

```python
sns.barplot(data=df,x='Family Size',y='Survived')
```

```python
df.head()
```

```python
def fam_type(num):
    if num>=1 and num<=3:
        return 'Small Family'
    elif num>=3 and num<=5:
        return 'Medium Family'
    else:
        return 'Large Family'
```

```python
df['Family Type']=df['Family Size'].apply(fam_type)
```

```python
df.head()
```

```python
df['Family Type'].value_counts()
```

```python
family = df['Family Type'].reset_index(drop=True)
survived = df['Survived'].reset_index(drop=True)

sns.heatmap(pd.crosstab(family, survived,normalize=True)*100)
```
